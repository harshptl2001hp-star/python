```python
# Indexing and slicing

items = ["a", "b", "c", "d", "e"]

# Indexing
print(items[0])    # first item
print(items[-1])   # last item

# Slicing: items[start:stop:step]
print(items[1:4])  # ['b', 'c', 'd']
print(items[:3])   # ['a', 'b', 'c']
print(items[2:])   # ['c', 'd', 'e']
print(items[::2])  # ['a', 'c', 'e']
print(items[::-1]) # reverse list
```
